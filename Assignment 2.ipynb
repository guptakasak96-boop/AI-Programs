```python
import numpy as np
```

```python
data={
    'singer':['Shreya Ghoshal','Kishore Kumar','Lata Mangeshkar','Arijit Singh','Sonu Nigam'],'Active Year':[20,40,70,15,30],'Number_of_Songs':[700,800,1000,500,600],'Awards_Won':[60,80,100,50,40]}
```

```python
singers_df = pd.DataFrame(data)
```

```python
singers_df
```

```python
singers_df.describe()
```

```python
max_songs = singers_df['Number_of_Songs'].max()
min_songs = singers_df['Number_of_Songs'].min()
mean_awards = singers_df['Awards_Won'].mean()
median_years = singers_df['Active Year'].median()
print(f"Maximun number of songs: {max_songs}")
print(f"Minimun number of songs: {min_songs}")
print(f"Average awards won: {mean_awards:.2f}")
print(f"Median years active: {median_years}")
```

```python
sorted_singers = singers_df.sort_values(by='Awards_Won',ascending=False)
sorted_singers
```

```python
singers_with_many_awards=singers_df[singers_df['Awards_Won']>50]
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(6,5))
plt.bar(singers_df['singer'], singers_df['Number_of_Songs'], color='violet')
plt.title('Number of Songs by Indian Singers')
plt.xlabel('Singer')
plt.ylabel('Number of Songs')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.show()
```
