```python
import pandas as pd
dataset_url = "movie.csv"
df = pd.read_csv(dataset_url)
```

```python
print("Dataset Overview:")
print(df.head())
```

```python
print("\nDataset Info:")
print(df.info())
print("\nDataset Statistics:")
print(df.describe())
```

```python
df.describe()
```

```python
sorted_df = df.sort_values(by="hitFlop",ascending=False)
sorted_df
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,5))
plt.bar(df["releaseYear"],df["hitFlop"])
plt.title("Types of movies")
plt.xlabel("ReleaseYear")
plt.ylabel("hit or flop movies")
plt.xticks(rotation=45)
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(7,7))
plt.pie(df["releaseYear"], labels=df["title"],autopct="%1.1f%%",startangle=90)
plt.title("Movie's over years")
plt.show()
```
